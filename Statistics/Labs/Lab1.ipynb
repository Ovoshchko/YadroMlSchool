$$
P(\xi = 1) = \frac{1}{3} + \frac{1}{6} = \frac{1}{2}

$$
P(\eta = 0) = \frac{1}{12} + \frac{1}{6} = \frac{6}{24} = \frac{1}{4}

$$
P(\eta = 1) = \frac{7}{24} + 0 = \frac{7}{24}

## Задание 1.2.3

Таблица совместного распределения $( \xi_1 ) и ( \xi_2 )$

| \( $\xi_1$ \) | 1       | 2       | 3       | 4       |
|---------------|---------|---------|---------|---------|
| **1**         | 1/16    | 1/16    | 1/16    | 1/16    |
| **2**         | 1/16    | 1/16    | 1/16    | 1/16    |
| **3**         | 1/16    | 1/16    | 1/16    | 1/16    |
| **4**         | 1/16    | 1/16    | 1/16    | 1/16    |

Каждая ячейка таблицы содержит вероятность $( P(\xi_1, \xi_2) = \frac{1}{16} )$.

Маргинальные распределения

$$
P(\xi_1 = k) = \frac{1}{16} + \frac{1}{16} + \frac{1}{16} + \frac{1}{16} = \frac{4}{16} = \frac{1}{4}
$$

Аналогично:

$$
P(\xi_2 = k) = \frac{1}{16} + \frac{1}{16} + \frac{1}{16} + \frac{1}{16} = \frac{4}{16} = \frac{1}{4}
$$

Математическое ожидание

$$
E(\xi_1) = E(\xi_2) = 1 \cdot \frac{1}{4} + 2 \cdot \frac{1}{4} + 3 \cdot \frac{1}{4} + 4 \cdot \frac{1}{4} = \frac{10}{4} = 2.5
$$

Ковариация

Сначала вычислим $( E(\xi_1 \xi_2) )$:

$$
E(\xi_1 \xi_2) = \frac{1}{16}(1\cdot1 + 1\cdot2 + 1\cdot3 + 1\cdot4 + 2\cdot1 + 2\cdot2 + 2\cdot3 + 2\cdot4 + 3\cdot1 + 3\cdot2 + 3\cdot3 + 3\cdot4 + 4\cdot1 + 4\cdot2 + 4\cdot3 + 4\cdot4) = 7.5
$$

Теперь находим ковариацию:

$$
\text{Cov}(\xi_1, \xi_2) = E(\xi_1 \xi_2) - E(\xi_1)E(\xi_2) = 7.5 - 6.25 = 1.25
$$

Сначала найдем дисперсию $( \text{Var}(\xi_1) )$:

$$
\text{Var}(\xi_1) = E(\xi_1^2) - (E(\xi_1))^2
$$

Где:

$$
E(\xi_1^2) = 1^2 \cdot \frac{1}{4} + 2^2 \cdot \frac{1}{4} + 3^2 \cdot \frac{1}{4} + 4^2 \cdot \frac{1}{4} = \frac{1 + 4 + 9 + 16}{4} = \frac{30}{4} = 7.5
$$

Таким образом:

$$
\text{Var}(\xi_1) = 7.5 - (2.5)^2 = 1.25
$$

Ковариционная матрица будет:

$$
\Sigma =
\begin{bmatrix}
1.25 & 1.25 \\
1.25 & 1.25
\end{bmatrix}
$$

Корреляционная матрица:

$$
R =
\begin{bmatrix}
1 & 1 \\
1 & 1
\end{bmatrix}
$$

Независимость
$$
P(\xi_1 = 1, \xi_2 = 1) = \frac{1}{16}
$$
$$
P(\xi_1 = 1) \cdot P(\xi_2 = 1) = \frac{1}{4} \cdot \frac{1}{4} = \frac{1}{16}
$$
Поскольку это равенство выполняется для всех возможных комбинаций, случайные величины $( \xi_1 ) и ( \xi_2 )$ **независимы**.

Некоррелированность

$$
\text{Cov}(\xi_1, \xi_2) = 1.25
$$

То есть величины **коррелированы**.


## Задание 1.2.4

Пусть \( $\xi \sim$ U(-$\pi, \pi$) \), тогда плотность распределения случайной величины \( $\xi$ \) равна:
$$
f(\xi) = \frac{1}{2\pi}, \quad \xi \in [-\pi, \pi].
$$

\( $\eta_1$ = $\cos \xi$ \) и \( $\eta_2 = \sin \xi$ \), ($\xi, \eta$), где $\eta = (\eta_1, \eta_2)$.

Математическое ожидание:
   $$
   \mathbb{E}[\xi] = \int_{-\pi}^{\pi} \xi \cdot \frac{1}{2\pi} d\xi = 0,
   $$
   поскольку интеграл нечётной функции на симметричном интервале равен нулю.

   $$
   \mathbb{E}[\cos \xi] = \int_{-\pi}^{\pi} \cos \xi \cdot \frac{1}{2\pi} d\xi = 0,
   $$
   $$
   \mathbb{E}[\sin \xi] = \int_{-\pi}^{\pi} \sin \xi \cdot \frac{1}{2\pi} d\xi = 0.
   $$
   Таким образом, математическое ожидание вектора $(\xi, \eta)$ равно $(0, 0, 0)$.

Ковариационная матрица:
   $$
   \text{Cov}(\xi, \xi) = \mathbb{E}[\xi^2] - (\mathbb{E}[\xi])^2 = \int_{-\pi}^{\pi} \xi^2 \cdot \frac{1}{2\pi} d\xi = \frac{\pi^2}{3},
   $$
   $$
   \text{Cov}(\eta_1, \eta_1) = \mathbb{E}[\cos^2 \xi] - (\mathbb{E}[\cos \xi])^2 = \frac{1}{2},
   $$
   $$
   \text{Cov}(\eta_2, \eta_2) = \mathbb{E}[\sin^2 \xi] - (\mathbb{E}[\sin \xi])^2 = \frac{1}{2},
   $$
   $$
   \text{Cov}(\eta_1, \eta_2) = \mathbb{E}[\cos \xi \sin \xi] - \mathbb{E}[\cos \xi] \mathbb{E}[\sin \xi] = 0.
   $$
   Также вычислим $\text{Cov}(\xi, \eta_1)$ и $\text{Cov}(\xi, \eta_2)$, которые также равны нулю, так как произведения $\xi, \cos \xi$ и $\xi, \sin \xi$ дают нечётные функции на симметричном интервале.

   Ковариационная матрица примет вид:
   $$
   \text{Cov}(\xi, \eta) = 
   \begin{bmatrix}
   \frac{\pi^2}{3} & 0 & 0 \\
   0 & \frac{1}{2} & 0 \\
   0 & 0 & \frac{1}{2}
   \end{bmatrix}.
   $$

Корреляционная матрица:

   Корреляционная матрица определяется как
   $$
   \text{Corr}(X, Y) = \frac{\text{Cov}(X, Y)}{\sqrt{\text{Var}(X) \text{Var}(Y)}}.
   $$
   Диагональные элементы будут равны 1, а все недиагональные элементы равны нулю.

Исследование на независимость и некоррелированность:

- **Независимость**: Поскольку \( $\eta_1 = \cos \xi $\) и \( $\eta_2 = \sin \xi$ \) зависят от \( $\xi$ \), векторы \( $\xi$ \) и \( $\eta$ \) не являются независимыми.

- **Некоррелированность**: Так как все ковариации между различными компонентами равны нулю, \( $\xi$ \), \( $\eta_1$ \), и \( $\eta_2$ \) являются некоррелированными.

Таким образом, переменные \( $\xi$ \) и \( $\eta$ \) некоррелированы, но не независимы.


## Задание 1.2.5

$ \xi \sim \text{Exp}(2) $ и $ \eta \sim U(0, 1) $. 

$$
f_Z(x) = \int_{-\infty}^{\infty} f_{\xi}(x - t) f_{\eta}(t) \, dt.
$$

### Плотности $ f_{\xi}(x) $ и $ f_{\eta}(x) $

1. Плотность экспоненциального распределения имеет вид:
   $$
   f_{\xi}(x) = 
   \begin{cases}
   2 e^{-2x}, & x \geq 0, \\
   0, & x < 0.
   \end{cases}
   $$

2. Плотность равномерного распределения имеет вид:
   $$
   f_{\eta}(t) = 
   \begin{cases}
   1, & 0 \leq t \leq 1, \\
   0, & \text{иначе}.
   \end{cases}
   $$

### Нахождение свёртки $ f_Z(x) $

Разделим вычисление на два случая:

1. **Для $ 0 \leq x < 1 $**:

   $$
   f_Z(x) = \int_0^x 2 e^{-2(x - t)} \cdot 1 \, dt = 2 e^{-2x} \int_0^x e^{2t} \, dt.
   $$

   После интегрирования:

   $$
   f_Z(x) = 2 e^{-2x} \left( \frac{e^{2x} - 1}{2} \right) = 1 - e^{-2x}.
   $$

2. **Для $ x \geq 1 $**:

   $$
   f_Z(x) = \int_{x-1}^1 2 e^{-2(x - t)} \cdot 1 \, dt = 2 e^{-2x} \int_{x-1}^1 e^{2t} \, dt.
   $$

   После интегрирования:

   $$
   f_Z(x) = 2 e^{-2x} \left( \frac{e^2 - e^{2(x-1)}}{2} \right) = e^{2(1 - x)} - e^{-2x}.
   $$

### Ответ

Плотность распределения суммы $ Z = \xi + \eta $ равна:

$$
f_Z(x) = 
\begin{cases}
e^{-2x}(e^{2x} - 1), & 0 \leq x < 1, \\
e^{-2x}(e^{2} - 1), & x \geq 1.
\end{cases}
$$
