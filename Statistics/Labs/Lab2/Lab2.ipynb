### Задание 1

Дана плотность распределения случайной величины $ \xi $:
$$
f_{\xi}(x) = \theta^2 x e^{-\theta x}, \quad x \geq 0,
$$
где $ \theta > 0 $.

Математическое ожидание $ \mathbb{E}[\xi] $:

   $$
   \mathbb{E}[\xi] = \int_0^{\infty} x f_{\xi}(x) \, dx = \int_0^{\infty} x \theta^2 x e^{-\theta x} \, dx = \theta^2 \int_0^{\infty} x^2 e^{-\theta x} \, dx.
   $$

   $$
   \int_0^{\infty} x^n e^{-\theta x} \, dx = \frac{n!}{\theta^{n+1}}.
   $$
   
   $$
   \mathbb{E}[\xi] = \theta^2 \cdot \frac{2!}{\theta^3} = \frac{2}{\theta}.
   $$

Математическое ожидание квадрата $ \mathbb{E}[\xi^2] $:

   $$
   \mathbb{E}[\xi^2] = \int_0^{\infty} x^2 f_{\xi}(x) \, dx = \theta^2 \int_0^{\infty} x^3 e^{-\theta x} \, dx.
   $$

   $$
   \int_0^{\infty} x^3 e^{-\theta x} \, dx = \frac{3!}{\theta^4}.
   $$
   
   $$
   \mathbb{E}[\xi^2] = \theta^2 \cdot \frac{6}{\theta^4} = \frac{6}{\theta^2}.
   $$

Дисперсия $ \text{Var}(\xi) $:

   $$
   \text{Var}(\xi) = \mathbb{E}[\xi^2] - (\mathbb{E}[\xi])^2 = \frac{6}{\theta^2} - \left( \frac{2}{\theta} \right)^2 = \frac{6}{\theta^2} - \frac{4}{\theta^2} = \frac{2}{\theta^2}.
   $$


```python
import numpy as np
import matplotlib.pyplot as plt

theta_values = [0.5, 2, 8]
k_values = [2**i for i in range(4, 16)]

def analytical_values(theta):
    expected_value = 2 / theta
    expected_square = 6 / theta**2
    variance = 2 / theta**2
    return expected_value, variance, expected_square

def generate_sample(theta, size):
    return np.random.gamma(shape=2, scale=1/theta, size=size)

for theta in theta_values:
    analytical_mean, analytical_var, analytical_square_mean = analytical_values(theta)
    
    mean_estimates = []
    var_estimates = []
    square_mean_estimates = []
    
    for k in k_values:
        sample = generate_sample(theta, k)
        
        mean_estimate = np.mean(sample)
        mean_estimates.append(mean_estimate)
        
        var_estimate = np.var(sample)
        var_estimates.append(var_estimate)
        
        square_mean_estimate = np.mean(sample**2)
        square_mean_estimates.append(square_mean_estimate)
    
    plt.figure(figsize=(18, 6))
    
    plt.subplot(1, 3, 1)
    plt.plot(k_values, mean_estimates, label=r'Оценка $\mathbb{E}[\xi]$', marker='o')
    plt.axhline(y=analytical_mean, color='r', linestyle='--', label=r'Аналитическое $\mathbb{E}[\xi]$')
    plt.xscale('log', base=2)
    plt.xlabel('Размер выборки (k)')
    plt.ylabel('Математическое ожидание')
    plt.title(r'Оценка $\mathbb{E}[\xi]$ при $\theta = ' + str(theta) + '$')
    plt.legend()
    
    plt.subplot(1, 3, 2)
    plt.plot(k_values, var_estimates, label=r'Оценка Var($\xi$)', marker='o')
    plt.axhline(y=analytical_var, color='r', linestyle='--', label=r'Аналитическая Var($\xi$)')
    plt.xscale('log', base=2)
    plt.xlabel('Размер выборки (k)')
    plt.ylabel('Дисперсия')
    plt.title(r'Оценка Var($\xi$) при $\theta = ' + str(theta) + '$')
    plt.legend()
    
    plt.subplot(1, 3, 3)
    plt.plot(k_values, square_mean_estimates, label=r'Оценка $\mathbb{E}[\xi^2]$', marker='o')
    plt.axhline(y=analytical_square_mean, color='r', linestyle='--', label=r'Аналитическое $\mathbb{E}[\xi^2]$')
    plt.xscale('log', base=2)
    plt.xlabel('Размер выборки (k)')
    plt.ylabel('Математическое ожидание квадрата')
    plt.title(r'Оценка $\mathbb{E}[\xi^2]$ при $\theta = ' + str(theta) + '$')
    plt.legend()
    
    plt.tight_layout()
    plt.show()
```

### Задание 2

Аналитические вычисления

Для плотности распределения случайной величины \( $\xi$ \) заданной как:

$$
f_{\lambda, a}(x) =
\begin{cases}
\lambda e^{-\lambda (x - a)}, & x \geq a \\
0, & \text{иначе}
\end{cases}
$$

$$
f_{2, 2}(x) =
\begin{cases}
2 e^{-2 (x - 2)}, & x \geq 2 \\
0, & \text{иначе}
\end{cases}
$$

Мода данной функции плотности находится в точке \(x = a\):

$$
\text{Мода} = 2
$$

Математическое ожидание:

$$
E[\xi] = \int_2^\infty x \cdot 2 e^{-2(x - 2)} \, dx
$$

$$
E[\xi] = 2 \int_2^\infty x e^{-2x + 4} \, dx = 2 e^4 \int_2^\infty x e^{-2x} \, dx
$$

Находим:

$$
E[\xi] = 2.5
$$

Медиана:

$$
\int_a^m f_{\lambda, a}(x) \, dx = 0.5
$$

$$
\int_2^m 2 e^{-2(x - 2)} \, dx = 0.5
$$

$$
1 - e^{-2(m - 2)} = 0.5 \Rightarrow e^{-2(m - 2)} = 0.5 \Rightarrow m - 2 = \frac{1}{2} \ln(2) \Rightarrow m = 2 + \frac{1}{2} \ln(2) \approx 2.3466
$$

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import expon

lambda_param = 2
a_param = 2

large_sample = np.random.exponential(1/lambda_param, 10000) + a_param
small_sample = np.random.exponential(1/lambda_param, 20) + a_param

def statistics(sample):
    mode = sample[np.argmax(np.bincount(sample.astype(int)))]
    mean = np.mean(sample)
    median = np.median(sample)
    return mode, mean, median

large_mode, large_mean, large_median = statistics(large_sample)
small_mode, small_mean, small_median = statistics(small_sample)

def plot_samples(sample, mode, mean, median, title):
    plt.figure(figsize=(12, 6))
    
    plt.subplot(1, 2, 1)
    plt.hist(sample, bins=30, density=True, alpha=0.5, color='g', label='Гистограмма')
    
    plt.axvline(mode, color='r', linestyle='dashed', linewidth=2, label='Мода')
    plt.axvline(mean, color='b', linestyle='dashed', linewidth=2, label='Математическое ожидание')
    plt.axvline(median, color='y', linestyle='dashed', linewidth=2, label='Медиана')
    
    plt.title(title)
    plt.legend()
    plt.grid(True)
    
    plt.subplot(1, 2, 2)
    sns.kdeplot(sample, bw_adjust=0.5, fill=True)
    
    plt.axvline(mode, color='r', linestyle='dashed', linewidth=2, label='Мода')
    plt.axvline(mean, color='b', linestyle='dashed', linewidth=2, label='Математическое ожидание')
    plt.axvline(median, color='y', linestyle='dashed', linewidth=2, label='Медиана')
    
    plt.title(title + ' (KDE)')
    plt.legend()
    plt.grid(True)

    plt.show()

plot_samples(large_sample, large_mode, large_mean, large_median, 'Большая выборка')
plot_samples(small_sample, small_mode, small_mean, small_median, 'Маленькая выборка')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

lambda_param = 2
a_param = 2

def statistics(sample):
    mode = sample[np.argmax(np.bincount(sample.astype(int)))]
    mean = np.mean(sample)
    median = np.median(sample)
    return mode, mean, median

sizes = [10, 1000, 100000, 10000000, 100000000]
results = []

for size in sizes:
    sample = np.random.exponential(1/lambda_param, size) + a_param
    mode, mean, median = statistics(sample)
    results.append((size, mode, mean, median))

modes, means, medians = zip(*[(res[1], res[2], res[3]) for res in results])
x_labels = sizes

plt.figure(figsize=(12, 6))
plt.plot(x_labels, modes, marker='o', label='Мода', color='r')
plt.plot(x_labels, means, marker='o', label='Математическое ожидание', color='b')
plt.plot(x_labels, medians, marker='o', label='Медиана', color='y')

plt.xscale('log')
plt.xlabel('Размер выборки')
plt.ylabel('Статистика')
plt.title('Статистика')
plt.legend()
plt.grid()
plt.show()
```

### Задание 3

$$
\begin{array}{c|ccc}
\eta & 1 & 2 & 3 \\
\hline
-1 & \frac{2}{5 \cdot 2} & \frac{2}{5 \cdot 4} & \frac{2}{5 \cdot 8} \\
0 & \frac{1}{5 \cdot 2} & \frac{1}{5 \cdot 4} & \frac{1}{5 \cdot 8} \\
1 & \frac{2}{5 \cdot 2} & \frac{2}{5 \cdot 4} & \frac{2}{5 \cdot 8} \\
\end{array}
$$

Математическое ожидание величины $\xi$ рассчитывается по формуле:

$$
E[\xi] = \sum_{\xi, \eta} \xi \cdot P(\xi, \eta)
$$

$$
E[\xi] = \sum_{\xi, \eta} \eta \cdot P(\xi, \eta)
$$


Для каждой $\eta$, вероятность $ P(\xi, \eta)$ уменьшается как $\frac{1}{2^\eta}$. Таким образом:

$$
E[\xi] = (-1) \cdot \left( \frac{2}{5} \sum_{\eta=1}^{\infty} \frac{1}{2^\eta} \right) + 0 \cdot \left( \frac{1}{5} \sum_{\eta=1}^{\infty} \frac{1}{2^\eta} \right) + 1 \cdot \left( \frac{2}{5} \sum_{\eta=1}^{\infty} \frac{1}{2^\eta} \right)
$$

Сумма геометрической прогрессии:

$$
\sum_{\eta=1}^{\infty} \frac{1}{2^\eta}
$$

Сумма бесконечной геометрической прогрессии может быть вычислена по следующей формуле, если знаменатель прогрессии \( r \) удовлетворяет условию \( |r| < 1 \):

$$
S = \frac{a}{1 - r}
$$

Где:
- $S$ — сумма прогрессии,
- $a$ — первый член прогрессии,
- $ r$ — знаменатель прогрессии (отношение каждого последующего члена к предыдущему).

В нашем случае подставим значения:

- $ a = \frac{1}{2} $
- $ r = \frac{1}{2} $


$$
S = \frac{\frac{1}{2}}{1 - \frac{1}{2}} = \frac{\frac{1}{2}}{\frac{1}{2}} = 1
$$

Получаем:

$$
\sum_{\eta=1}^{\infty} \frac{1}{2^\eta}
$$

$$
E[\xi] = (-1) \cdot \frac{2}{5} + 0\cdot \frac{1}{5} + 1 \cdot \frac{2}{5} = 0
$$


$$
E[\eta] = \sum_{\eta=1}^{\infty} \eta \cdot \left( \frac{2}{5} \cdot \frac{1}{2^\eta} + \frac{1}{5} \cdot \frac{1}{2^\eta} + \frac{2}{5} \cdot \frac{1}{2^\eta} \right)
$$

Это выражение можно свести к:

$$
E[\eta] = \sum_{\eta=1}^{\infty} \eta \cdot \frac{1}{2^\eta}
$$


Чтобы получить сумму $ \sum_{n=1}^{\infty} n r^n $, продифференцируем по $r$:

$$
\frac{dS}{dr} = \sum_{n=1}^{\infty} n r^{n-1} = \frac{1}{(1 - r)^2}
$$

$$
r \cdot \frac{dS}{dr} = \sum_{n=1}^{\infty} n r^n = r \cdot \frac{1}{(1 - r)^2}
$$

Таким образом, итоговая формула для суммы:

$$
\sum_{n=1}^{\infty} n r^n = \frac{r}{(1 - r)^2}
$$


Для $r = \frac{1}{2}$, подставляем в формулу:

$$
\sum_{\eta=1}^{\infty} \eta \cdot \frac{1}{2^\eta} = \frac{\frac{1}{2}}{(1 - \frac{1}{2})^2} = \frac{\frac{1}{2}}{\left( \frac{1}{2} \right)^2} = \frac{\frac{1}{2}}{\frac{1}{4}} = 2
$$

$$
E[\eta] = 2
$$

$$
E[\xi \eta] = \sum_{\xi, \eta} \xi \eta \cdot P(\xi, \eta)
$$

$$
E[\xi \eta] = (-1) \cdot \sum_{\eta=1}^{\infty} \eta \cdot \frac{2}{5} \cdot \frac{1}{2^\eta} + 0 \cdot \sum_{\eta=1}^{\infty} \eta \cdot \frac{1}{5} \cdot \frac{1}{2^\eta} + 1 \cdot \sum_{\eta=1}^{\infty} \eta \cdot \frac{2}{5} \cdot \frac{1}{2^\eta}
$$

$$
E[\xi \eta] = (-1) \cdot \frac{2}{5} \cdot 2 + 1 \cdot \frac{2}{5} \cdot 2 = 0
$$


Ковариация между величинами $\xi$ и $\eta$ рассчитывается по формуле:

$$
\text{Cov}(\xi, \eta) = E[\xi \eta] - E[\xi] \cdot E[\eta]
$$


$$
\text{Cov}(\xi, \eta) = 0 - 0 \cdot 2 = 0
$$


Дисперсии величин $\xi$ и $\eta$ рассчитываются по формулам:

$$
\text{D}[\xi] = E[\xi^2] - (E[\xi])^2
$$

$$
\text{D}[\eta] = E[\eta^2] - (E[\eta])^2
$$


$$
E[\xi^2] = (-1)^2 \cdot \frac{2}{5} \cdot 1 + 0^2 \cdot \frac{1}{5} \cdot 1 + 1^2 \cdot \frac{2}{5} \cdot 1 = \frac{4}{5}
$$

$$
E[\eta^2] = \sum_{\eta=1}^{\infty} \eta^2 \cdot \frac{1}{2^\eta}
$$


Для этого воспользуемся известной формулой для суммы, которая включает $n^2 r^n$:

$$
\sum_{n=1}^{\infty} n^2 r^n = \frac{r(1+r)}{(1-r)^3}
$$

Подставляем $ r = \frac{1}{2} $:

$$
E[\eta^2] = \sum_{\eta=1}^{\infty} \eta^2 \cdot \frac{1}{2^\eta} = \frac{\frac{1}{2}(1+\frac{1}{2})}{(1 - \frac{1}{2})^3} = \frac{\frac{1}{2} \cdot \frac{3}{2}}{\left( \frac{1}{2} \right)^3}
$$

$$
E[\eta^2] = \frac{\frac{3}{4}}{\frac{1}{8}} = \frac{3}{4} \cdot 8 = 6
$$

Вычисляем дисперсию

$$
\text{D}[\xi] = \frac{4}{5} - 0^2 = \frac{4}{5}
$$


$$
\text{D}[\eta] = 6 - 2^2 = 2
$$


Формула корреляции

$$
\rho(\xi, \eta) = \frac{\text{Cov}(\xi, \eta)}{\sqrt{\text{Var}[\xi] \cdot \text{Var}[\eta]}}
$$


$$
\rho(\xi, \eta) = \frac{0}{\sqrt{\frac{4}{5} \cdot 2}} = 0
$$


```python
import numpy as np

xi_values = np.array([-1, 0, 1])
max_eta = 200

def probability(xi, eta):
    if xi == -1 or xi == 1:
        return 2/5 * 1/2**eta
    elif xi == 0:
        return 1/5 * 1/2**eta

probabilities = np.array([[probability(xi, eta) for eta in range(1, max_eta+1)] for xi in xi_values])

assert np.isclose(probabilities.sum(), 1), "Сумма вероятностей не равна 1!"

xi_eta_pairs = [(xi, eta) for xi in xi_values for eta in range(1, max_eta+1)]
flattened_probabilities = probabilities.flatten()
samples = np.random.choice(len(xi_eta_pairs), p=flattened_probabilities, size=1000)
xi_samples = [xi_eta_pairs[sample][0] for sample in samples]
eta_samples = [xi_eta_pairs[sample][1] for sample in samples]
correlation = np.corrcoef(xi_samples, eta_samples)[0, 1]

print("Корреляция между ξ и η:", correlation)
```
